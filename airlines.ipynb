```python
import kagglehub
from pathlib import Path
import pandas as pd
import numpy as np

# Download directly from Kaggle
download_path = kagglehub.dataset_download(
    "waleedfaheem/airline-route-profitability-and-cost-analysis"
)

print("Dataset downloaded successfully!")
print("Location:", download_path)

# Find CSV files
csv_files = list(Path(download_path).rglob("*.csv"))

for file in csv_files:
    print(file)
```

```python
if not csv_files:
    raise FileNotFoundError("No CSV file found in the downloaded dataset.")

# Load the first CSV file
df = pd.read_csv(csv_files[0])

print("Dataset loaded successfully!")
print("Shape:", df.shape)

display(df.head())
```

```python
df.columns = (
    df.columns
    .str.strip()
    .str.replace(" ", "_")
)

print(df.columns.tolist())
```

```python
before = len(df)

df = df.drop_duplicates().copy()

print("Duplicates removed:", before - len(df))
```

```python
if "Flight_Date" in df.columns:
    df["Flight_Date"] = pd.to_datetime(
        df["Flight_Date"], errors="coerce"
    )

numeric_columns = [
    "Aircraft_Capacity",
    "Passengers",
    "Load_Factor",
    "Flight_Hours",
    "Ticket_Revenue",
    "Ancillary_Revenue",
    "Total_Revenue",
    "Total_Cost",
    "Profit",
    "Profit_Margin"
]

for col in numeric_columns:
    if col in df.columns:
        df[col] = pd.to_numeric(df[col], errors="coerce")

print("Data types corrected.")
df.info()
```

```python
print("Missing values before cleaning:")
print(df.isnull().sum())

required_columns = [
    "Flight_Number",
    "Flight_Date",
    "Origin",
    "Destination"
]

available_columns = [
    col for col in required_columns if col in df.columns
]

df = df.dropna(subset=available_columns).copy()

print("Shape after removing rows missing essential fields:", df.shape)
```

```python
if {"Passengers", "Aircraft_Capacity"}.issubset(df.columns):
    df["Load_Factor"] = np.where(
        df["Aircraft_Capacity"] > 0,
        df["Passengers"] / df["Aircraft_Capacity"],
        np.nan
    )

display(df[["Passengers", "Aircraft_Capacity", "Load_Factor"]].head())
```

```python
output_file = Path.cwd() / "cleaned_airline_data.csv"

df.to_csv(output_file, index=False)

print("Cleaned dataset saved at:")
print(output_file.resolve())

print("Final dataset shape:", df.shape)
```

```python

import pandas as pd
import numpy as np

np.random.seed(42)

# Create 500 sample flight records
n = 500

df_delay = pd.DataFrame({
    "Flight_ID": range(1, n + 1),
    "Flight_Date": pd.date_range(
        start="2026-01-01",
        periods=n,
        freq="D"
    ),
    "Origin": ["DXB"] * n,
    "Destination": np.random.choice(
        ["JFK", "LHR", "BOM", "SIN", "DEL", "CDG"],
        n
    ),
    "Arrival_Delay": np.random.normal(
        loc=10, scale=20, size=n
    ).round().astype(int)
})

# Simulate cancelled flights
df_delay["Cancelled"] = np.random.choice(
    [0, 1], size=n, p=[0.97, 0.03]
)

# Missing delay for cancelled flights
df_delay.loc[df_delay["Cancelled"] == 1, "Arrival_Delay"] = np.nan

# Save the dataset
df_delay.to_csv("simulated_flight_delays.csv", index=False)

print("Sample delay data created!")
display(df_delay.head(10))
```
